```python
# 🟢 Setup and Required Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.integrate import odeint
import seaborn as sns
import jinja2

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
%matplotlib inline

print("All libraries imported successfully!")
```

```python
def sird_model(y, t, beta, gamma, mu):
    """
    SIRD model with death compartment

    TODO: Adapt the SIR model to include deaths

    Additional parameter:
    - mu: mortality rate

    New compartment:
    - D: Deaths
    """
    S, I, R, D = y  # Note: now 4 compartments
    N = S + I + R + D

    # Equations

    dSdt = -beta * S * I / N
    dIdt = beta * S * I / N - gamma * I - mu * I
    dRdt = gamma * I
    dDdt = mu * I

    return dSdt, dIdt, dRdt, dDdt
```

```python
def run_sird_simulation(beta, gamma, mu, N=1000, I0=10, days=150, plot=False):


    S0 = N - I0
    R0 = 0
    D0 = 0
    y0 = [S0, I0, R0, D0]

    # Time points (same as before)
    # Note: days - 1 as the endpoint so that t[i] == i exactly
    t = np.linspace(0, days - 1, days)

    sol = odeint(sird_model, y0, t, args=(beta, gamma, mu))
    S, I, R, D = sol.T

    if plot:
      # 4 lines plots
        plt.figure(figsize=(10, 6))
        plt.plot(t, S, 'b-', label='Susceptible', linewidth=2)
        plt.plot(t, I, 'r-', label='Infectious', linewidth=2)
        plt.plot(t, R, 'g-', label='Recovered', linewidth=2)
        plt.plot(t, D, 'k-', label='Deaths', linewidth=2)

        plt.xlabel('Time (days)')
        plt.ylabel('Population')
        plt.title(f'SIRD Model: β={beta}, γ={gamma}, μ={mu}')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()

    return t, S, I, R, D
```

## Part 1: Parameter analysis function.

```python
def analyze_recovery_rates(beta, gamma_values, mu, N, I0, simulation_days, title):
    """
    Analyze epidemic outcomes for different recovery rates.

    Parameters:
    -----------
    beta : float
        Transmission rate
    mu : float
        Mortality rate
    N : int
        Total population
    I0 : int
        Initial infected individuals
    simulation_days : int
        Simulation duration in days

    Returns:
    --------
    pandas.DataFrame
        Results summary for each recovery rate
    """
    # Your implementation here
    results = []
    fig, ax = plt.subplots(figsize=(10, 6))

    for i,gamma in enumerate(gamma_values):
        t, S, I, R, D = run_sird_simulation(beta, gamma, mu, N, I0, simulation_days)

        # Store results
        results.append({
            "Recovery Rate (γ)": gamma,
            "Peak Infected": np.max(I),
            "Peak Infected Day": np.argmax(I),
            "Total Deaths": D[-1],
            "R₀": beta / gamma
        })

        # Plot epidemic curve
        ax.plot(t, I, label=f'γ={gamma:.2f}')

    ax.set_xlabel('Time (days)')
    ax.set_ylabel('Number of Infectious Individuals')
    ax.set_title(f'Effect of Recovery Rate on Epidemic Dynamics - {title}')
    ax.legend(title='Recovery Rate (γ)', frameon=False)
    ax.grid(alpha=0.2)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    fig.tight_layout()

    df = pd.DataFrame(results)
    df = df.style.format({
        "Recovery Rate (γ)": "{:.2f}",
        "Peak Infected": "{:.0f}",
        "Peak Infected Day": "{:.0f}",
        "Total Deaths": "{:.0f}",
        "R₀": "{:.2f}"
    })

    return df, fig, ax

```

### 3.3 Real-world application


In a large hospital there could be several problems, such as long wait times, which delay treatment and prolong recovery, crowded waiting rooms, which can increase the transmission rate, overworked staff, which reduces care quality and can mean worse outcomes, and insufficient or inefficient resources, which create limited access to specialized care, leading to slower recovery. Therefore, it is better to have a health system with multiple centers, each with different facilities for different severity levels. For example, testing centers for mild cases, primary care clinics for moderate cases and hospitals for severe cases. Additionally, there could be centers for different specialized treatments or temporary ones in case of an epidemic. By having these changes, the waiting time could lower, staff could be able to check on patients more and treatment would start earlier. This would, in conclusion, increase recovery rate by reducing the waiting time and improving care quality and reduce transmission rate by decreasing crowding.

In other words, the takeaway is; If we want lower infection peaks, reduce death numbers and shorten epidemic durations, then practices that increase the recovery rate are the most effective. A lower transmission rate could be more affective but that should be considered after optimizing recovery rate.

## Appendix - Fitting a curve to our results

```python
from scipy.optimize import curve_fit

gamma_values = np.linspace(0.05, 0.9, 100)  # Testing bigger range and more points for better curve fitting
# High transmission scenario
df1, fig1, ax1 = analyze_recovery_rates(beta=0.4, gamma_values=gamma_values, mu=0.02, N=1000, I0=5, simulation_days=200, title="figure1: High Transmission Scenario")
# Low transmission scenario
df2, fig2, ax2 = analyze_recovery_rates(beta=0.2, gamma_values=gamma_values, mu=0.02, N=1000, I0=5, simulation_days=200, title="figure2: Low Transmission Scenario")
```

```python
def exp_model(gamma, a, b):
    return a * np.exp(b * gamma)

def log_model(gamma, a, b):
    return a + b*np.log(gamma)

def r_squared(y_true, y_pred):
    """
    Calculate the coefficient of determination (R²) for model fit.

    Parameters:
    -----------
    y_true : True values
    y_pred : Predicted values from the model

    Returns:
    --------
    float
        R² value indicating goodness of fit
    """
    ss_res = np.sum((y_true - y_pred) ** 2) # Residual sum of squares
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2) # TotaL sum of squares
    return 1 - (ss_res / ss_tot)
```

### High Transmission Curve Fit

First we fit an exponential and a logarithmic curve to see which one is a better fit for our results.

```python
x = df1.data["Recovery Rate (γ)"].values
y = df1.data["Peak Infected"].values

opt_params_exp, _ = curve_fit(exp_model, x, y)
r2_exp = r_squared(y, exp_model(x, *opt_params_exp))

opt_params_log, _ = curve_fit(log_model, x, y)
r2_log = r_squared(y, log_model(x, *opt_params_log))

print(f"Exponential Model Parameters: a={opt_params_exp[0]:.4f}, b={opt_params_exp[1]:.4f}, R²={r2_exp:.4f}")
print(f"Logarithmic Model Parameters: a={opt_params_log[0]:.4f}, b={opt_params_log[1]:.4f}, R²={r2_log:.4f}")
```

Note: Exponential clearly better fit.

```python
def plot_fits(x, y, opt_params_exp, opt_params_log, r2_exp, r2_log):
    """ 
    Plot observed peak infection data alongside exponential and logarithmic model fits.
    Parameters:
    ---------- 
    x : array
        Recovery rate (gamma) values used as the independent variable.

    y : array
        Observed peak number of infected individuals. 

    opt_params_exp : array
                    Optimized parameters for the exponential model. 

    opt_params_log : array 
                    Optimized parameters for the logarithmic model. 

    r2_exp : float 
            R-squared value of the exponential model fit. 

    r2_log : float 
            R-squared value of the logarithmic model fit. 
    
    Returns: 
    ------- 
    None. Displays the plot. """
    x_smooth = np.linspace(x.min(), x.max(), 200)
    y_exp_smooth = exp_model(x_smooth, *opt_params_exp)
    y_log_smooth = log_model(x_smooth, *opt_params_log)

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.plot(x, y, color="blue", linewidth=2, label="Observed")
    ax.plot(
        x_smooth,
        y_exp_smooth,
        "--",
        label=f"Exponential fit (R²={r2_exp:.4f})"
    )
    ax.plot(
        x_smooth,
        y_log_smooth,
        ":",
        label=f"Logarithmic fit (R²={r2_log:.4f})"
    )

    ax.set_title(
        "Peak Infected vs. Recovery Rate: Exponential vs Logarithmic Fit",
        fontweight="bold"
    )
    ax.set_xlabel("Recovery Rate (γ)")
    ax.set_ylabel("Peak Infected")
    ax.legend(frameon=False)
    ax.grid(alpha=0.3)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
plot_fits(x, y, opt_params_exp, opt_params_log, r2_exp, r2_log)
```

```python
def plot_residuals(x, y, opt_params_exp, opt_params_log):
    """
    Plot residuals for the exponential and logarithmic model fits.

    Residuals are calculated as the difference between the observed
    and predicted peak number of infected individuals.

    Parameters
    ----------
    x : array
        Recovery rate (gamma) values used as the independent variable.

    y : array
        Observed peak number of infected individuals.

    opt_params_exp : array
        Optimized parameters for the exponential model.

    opt_params_log : array
        Optimized parameters for the logarithmic model.

    Returns
    -------
    None
        Displays the residual plot.
    """
    resid_exp = y - exp_model(x, *opt_params_exp)
    resid_log = y - log_model(x, *opt_params_log)

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.axhline(0, color="gray", linewidth=1, zorder=1)
    ax.plot(x, resid_exp, "o--", label="Exponential residuals")
    ax.plot(x, resid_log, "o--", label="Logarithmic residuals")

    ax.set_title("Residuals vs. Recovery Rate (γ)", fontweight="bold")
    ax.set_xlabel("Recovery Rate (γ)")
    ax.set_ylabel("Residual (Observed − Predicted)")
    ax.legend(frameon=False)
    ax.grid(alpha=0.3)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()
    plt.show()


plot_residuals(x, y, opt_params_exp, opt_params_log)
```

Exponential approaches near 0 residual.
